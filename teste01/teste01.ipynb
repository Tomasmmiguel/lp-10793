### 5. Completar — lista de preços [20 pontos]

Substitui cada ____ pelo que falta, seguindo o comentário de cada linha. Não mudes mais nada.

Depois de completo, o programa mostra:

```text

```

````python
precos = [4.5, 2.0, 4.0]
precos.append(3.0)            # acrescenta 3.0 ao fim da lista
precos.remove(4.5)            # remove o 4.5 da lista
precos.sort()                 # ordena do mais barato para o mais caro
for i, preco in enumerate(precos, 1):   # numera a partir de 1
    print(f"{i}. {preco:.2f} €")
print(f"Total: {sum(precos):.2f} €")   # soma todos os preços
````

### 7. Corrigir — função que calcula a média [20 pontos]

A função media devia devolver a média de uma lista de notas, mas o código tem 3 erros. Executa, lê a última linha de cada mensagem de erro e corrige um erro de cada vez.

Em cada linha corrigida, acrescenta no fim um comentário # a dizer o que estava errado. Depois de corrigido, o programa mostra:

```text
Média: 14.0
```

````python
def media(notas):
    total = 0
    for nota in notas:
        total += nota
    return total / len(notas)

resultado = int(media([17, 14, 11]))
print("Média: " + str(resultado))
````

### 11. Problema — produtos mais vendidos [20 pontos]

O dicionário vendas, que já está na célula de código, guarda quantas unidades se venderam de cada produto no bar da escola. Resolve uma alínea de cada vez, por baixo do comentário respetivo.

**a)** Define a função mais_vendidos(vendas, minimo=60), que devolve uma lista com os nomes dos produtos com vendas maiores ou iguais a minimo. Passos: cria uma lista vazia; percorre o dicionário com items(); se as vendas do produto forem maiores ou iguais a minimo, acrescenta o produto à lista com append(); no fim, devolve a lista com return. [12 pontos]

**b)** Chama a função só com vendas e mostra a lista devolvida: ['iogurte', 'café', 'sandes'] [3 pontos]

**c)** Chama a função com o argumento nomeado minimo=120 e mostra quantos produtos tem a lista devolvida, com len(): Com 120 ou mais: 2 [5 pontos]

````python
vendas = {"iogurte": 125, "café": 150, "chá": 40, "água": 10, "sandes": 80}

# a)
def mais_vendido(vendas,minimo=60):
    lista = []
    for produto, quantidade in vendas.items():
        if quantidade >= minimo:
            lista.append(produto)
    return lista

# b)
print(mais_vendido(vendas))

# c)
resultado = mais_vendido(vendas, minimo=120)
print(f"Com mais de 120 vendas: {len(resultado)}")
````
