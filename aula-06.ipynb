# Exercício 3: completa os espaços

```python
temperatura = int(input("Temperatura em graus: ")) 
if temperatura >= 30: print("Está calor") 
elif temperatura >= 20: print("Está agradável") 
elif temperatura >5: print("Está frio")
else: print("Está muito frio")
```

# Exercício 4: altera um exemplo que funciona

# Atividade: bilheteira do cinema

```python
idade = int(input("Idade: "))
tem_cartao = True
if idade < 12: preco = 4.5
elif idade < 18: preco = 6.0
else: preco = 8.0
if tem_cartao and idade >= 18: preco -= 1
print(f"Preço do bilhete: {preco:.2f} €")
```
