# Exercício 2 — completa o código

```python
preco_refeicao = 2.50
nome = input("Nome: ")
refeicoes = int(input("Quantas refeições? "))
total =  refeicoes * preco_refeicao
print(f"{nome}, vais pagar {total:.2f} €.")
```
