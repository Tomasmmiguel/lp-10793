# 1. Que causa tem mais acidentes no total?

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["feridos"])

df_2024 = df[df["ano"] == 2024]
feridos_distrito = df_2024.groupby("distrito")["feridos"].sum()

print(feridos_distrito)
print("Distrito com mais feridos em 2024:", feridos_distrito.idxmax())

feridos_distrito.plot(kind="bar")
plt.title("Feridos por distrito em 2024")
plt.xlabel("Distrito")
plt.ylabel("Número de feridos")
plt.show()
```

# 2. Em que mês há mais acidentes?

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["feridos"])

df_2024 = df[df["ano"] == 2024]

acidentes_mes = df_2024.groupby("mes")["acidentes"].sum()
print(acidentes_mes)
print("Mês com mais acidentes:", acidentes_mes.idxmax())

acidentes_mes.plot(kind="line", marker="o")
plt.title("Acidentes por mês em 2024")
plt.xlabel("Mês")
plt.ylabel("Número de acidentes")
plt.xticks(range(1, 13))
plt.show()
```
