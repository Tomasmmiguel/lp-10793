# 1. Prever — atribuição composta 

```python
pontos = 10 
pontos += 5 
pontos *= 2 
pontos -= 4 
pontos //= 3 
print(pontos)
```

# 2. Prever — conversão de tipos 

```python
print(int("25") + 5) 
print("25" + "5") 
print(float("2.5") * 2) 
print(int(7.9)) 
print(str(3) * 3) 
```

# 3. Prever — métodos de texto 

```python
cidade = "  lousada  " 
print(len(cidade)) 
cidade = cidade.strip() 
print(len(cidade)) 
print(cidade.upper()) 
print(cidade.title()) 
print(cidade.count("a")) 
print(cidade.replace("a", "@"))
```

# 4. Completar — cartão da cantina

```python
saldo = 20.0 
saldo -= 2.5    # paga uma refeição 
saldo += 10     # carrega 10 € 
saldo -= 2.5    # paga outra refeição 
print(f"Saldo: {saldo:.2f} €") 
```

# 5. Completar — conversão de tipos 

```python
nome = input("Nome: ") 
cafes = int(input("Quantos cafés? ")) 
preco = float(input("Preço de um café (ex.: 1.5): ")) 
total = cafes * preco 
print(nome + " comprou " + int(cafes) + " cafés e pagou " + float(total) + " €.")
```

# 6. Alterar — um programa que funciona 

```python
nome = input("Nome completo: ") 
nome = nome.strip() 
nome = nome.title() 
print(nome) 
print("Tem", len(nome), "caracteres.")
print("Aparece a letra a", nome.count("a"), "vezes.")
```

# 7. Escrever de raiz — linha de um talão 

```python
produto = input("Produto: ")
preco = float(input("Preço: "))
quantidade = int(input("Quantidade: "))
total = preco * quantidade
print(f"{produto:<12}{total:>8.2f} €")
```

# 8. Extensão (opcional) — preços com vírgula 

```python
texto = input("Preço: ")
texto = texto.replace(",", ".")
preco = float(texto)
dobro = preco * 2
print(f"O dobro de {preco:.2f} € é {dobro:.2f} €.")
```
