# 1. Prever — atribuição composta 

```python
pontos = 10 
pontos += 5 
pontos *= 2 
pontos -= 4 
pontos //= 3 
print(pontos)
```

# 2. Prever — conversão de tipos 

```python
print(int("25") + 5) 
print("25" + "5") 
print(float("2.5") * 2) 
print(int(7.9)) 
print(str(3) * 3) 
```

# 3. Prever — métodos de texto 

```python
cidade = "  lousada  " 
print(len(cidade)) 
cidade = cidade.strip() 
print(len(cidade)) 
print(cidade.upper()) 
print(cidade.title()) 
print(cidade.count("a")) 
print(cidade.replace("a", "@"))
```

# 4. Completar — cartão da cantina

```python
saldo = 20.0 
saldo -= 2.5    # paga uma refeição 
saldo += 10     # carrega 10 € 
saldo -= 2.5    # paga outra refeição 
print(f"Saldo: {saldo:.2f} €") 
```

# 5. Completar — conversão de tipos 

```python
nome = input("Nome: ") 
cafes = int(input("Quantos cafés? ")) 
preco = float(input("Preço de um café (ex.: 1.5): ")) 
total = cafes * preco 
print(nome + " comprou " + int(cafes) + " cafés e pagou " + float(total) + " €.")
```
