# Exercício 1: completa e guarda a lista de compras

```python
with open("compras.txt", "w", encoding="utf-8") as f:
    f.write("pão\n")
    f.write("leite\n")
    f.write("ovos\n")
print("Lista guardada.")
print("compras.txt:")
with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        print(linha.strip())
```

# Exercício 2: prevê a leitura da lista de compras

```python
contador = 0
with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        contador += 1
        print(f"{contador}. {linha.strip()}")
print(f"Total: {contador} produtos")
```

# Exercício 3: acrescentar ou escrever de novo

```python
with open("compras.txt", "a", encoding="utf-8") as f:
    f.write("fruta\n")

with open("compras.txt", "r", encoding="utf-8") as f:
    for linha in f:
        print(linha.strip())
```

# Atividade: lista de tarefas guardada num ficheiro

```python
def mostrar_tarefas():
    try:
        with open("tarefas.txt", "r", encoding="utf-8") as f:
            numero = 0
            for linha in f:
                numero += 1
                print(f"{numero}. {linha.strip()}")
    except FileNotFoundError:
        print("Ainda não há tarefas.")
def acrescentar_tarefa(tarefa):
    with open("tarefas.txt", "a", encoding="utf-8") as f:
        f.write(tarefa + "\n")

mostrar_tarefas()
nova = input("Nova tarefa: ")
acrescentar_tarefa(nova)
mostrar_tarefas()
```
