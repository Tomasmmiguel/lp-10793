# Exercício 2: prevê o resultado

```python
import numpy as np
p = np.array([2, 4, 6, 8])
q = np.array([1, 1, 2, 2])
print(p - q)
print(p * q)
print(p / 2)
print(p[p > 4])
print((p + q) > 6)
```

# Exercício 3: completa as estatísticas

```python
import numpy as np

temp = np.array([14, 17, 12, 19, 21, 15, 14])
media = temp.mean()
maximo = temp.max()
posicao = np.argmax(temp)
frios = (temp < 15).sum()
print(media, maximo, posicao, frios)
```

# Atividade: as notas de uma turma com NumPy

```python
import numpy as np
notas = np.array([
    [12, 14, 11],
    [8, 10, 9],
    [16, 15, 18],
    [10, 12, 13],
    [5, 9, 7],
])

print(notas.shape)
print(np.round(notas.mean(axis=0), 1))
media = notas.mean(axis=1)
print(np.round(media, 2))
print(media.argmax() + 1)
print((media >= 10).sum())
```

# Desafio 1: valores fora do normal, sem ciclos

```python
import numpy as np
valores = np.array([52, 48, 50, 51, 49, 95, 50, 53])

z = (valores - valores.mean()) / valores.std()
print(np.round(z, 2))
print(valores[np.abs(z) > 1.5])
limpos = valores[np.abs(z) <= 1.5]
print(limpos.mean())
```

# Desafio 2: estimar o π com números aleatórios

```python
import math
import numpy as np

rng = np.random.default_rng(1)
for n in (1000, 100000, 1000000):
    x = rng.random(n)
    y = rng.random(n)
    dentro = x ** 2 + y ** 2 <= 1
    estimativa = 4 * dentro.mean()
    print(n, estimativa, round(abs(estimativa - math.pi), 4))

```
