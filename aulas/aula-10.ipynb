# Exercício 1: o que existe e onde

```python
def calcula_media(notas):
    soma_notas = sum(notas)
    quantidade = len(notas)
    return soma_notas / quantidade

m = calcula_media([12, 16, 14])
print(f"{m:.2f}")
```

# Exercício 2: outra função, o mesmo aplica_a_todos

```python
def quadrado(x):
    return x ** 2
def dobro(x):
    return x * 2
def aplica_a_todos(lista, funcao):
    return [funcao(x) for x in lista]
numeros = [1, 2, 3, 4]
print(aplica_a_todos(numeros, quadrado))
print(aplica_a_todos(numeros, dobro))
```

# Exercício 3: completa a compreensão

```python
cubos = [n ** 3 for n in range(1, 6)] # queremos [1, 8, 27, 64, 125]
impares = [n for n in range(1, 11) if n % 2 == 1] # queremos só os ímpares
print(cubos)
print(impares)
```

# Exercício 4: completa a compreensão de dicionário

```python
notas = {"Ana": 14, "Bruno": 9, "Carla": 18, "Duarte": 6}
reprovados = {nome: nota for nome, nota in 
notas.items() if nota < 10}
print(reprovados)
```

# Atividade: processador de notas

```python
def classifica(nota):
    if nota < 10:
        return "Reprovado"
    elif nota < 20:
        return "Aprovado"
    else:
        return "Nota invalida"

notas = {}

for i in range(3):
    nome = input("Nome do aluno: ")
    nota = float(input("Nota do aluno: "))
    notas[nome] = nota

resultado = {nome: classifica(nota) for nome, nota in notas.items()}

for nome, classificacao in resultado.items():
    print(f"{nome}: {classificacao}")

lista_aprovados = sorted([nome for nome, r in resultado.items() if r == "Aprovado"])
print("Alunos aprovados:", lista_aprovados)
```
