# Exercício 2: prevê e completa

```python
print(int("8") + 2) 
print(float("8") + 2) 
print(str(8) + "2") 
print(int(9.99)) 
print(float("9.99"))
```

```python
nome = input("Nome: ") 
idade = int(input("Idade: ")) 
altura = float(input("Altura em metros: ")) 
idade_daqui_a_10 = idade + 10 
print(nome + " daqui a 10 anos terá " + str(idade_daqui_a_10) + " anos.") 
print(nome + " mede " + str(altura) + " m.")
```

# Exercício 3: completa e prevê

```python
email = input("Email: ") 
email = email.strip() # tira os espaços do início e do fim 
email = email.lower() # passa tudo para minúsculas print(email) 
print(email)
print(len(email)) 
print(email.count(".")) 
print(email.replace("@", " (at) "))
```

# Exercício 4: completa a lista de preços

```python
produto1 = "Sopa" 
preco1 = 0.9 
produto2 = "Prato do dia" 
preco2 = 2.5 
print(f"{produto1:_<15} {preco1:.2f} €") 
print(f"{produto2:_<15} {preco2:.2f} €")
total = preco1 + preco2
print(f"{"-":-<15}")
print(f"{'Total':_<15} {total:.2f} €")
```
