# Exercício 1: prevê o resultado

```python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
```

# Exercício 2: completa as forma

```python
class Forma:
    def area(self):
        return 0

class Quadrado(Forma):
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

class Circulo(Forma):
    def __init__(self, raio):
        self.raio = raio

    def area(self):
        return 3.14 * self.raio ** 2

for f in [Quadrado(3), Circulo(2)]:
 print(f.area())
```

# Exercício 3: altera um exemplo que funciona

```python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4

    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")

class Mota(Veiculo):
    def __init__(self, marca):
        super().__init__(marca)
        self.rodas = 2

    def empinar(self):
        print(f"A {self.marca} empinou!")

m = Mota("Yamaha")
m.descrever()
m.empinar()
```

# Atividade: os pagamentos da empresa

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

    def pagamento(self):
        return self.salario

class Gestor(Funcionario):
    def __init__(self, nome, salario, bonus):
        super().__init__(nome, salario)
        self.bonus = bonus

    def pagamento(self):
        return super().pagamento() + self.bonus

class Estagiario(Funcionario):
    def pagamento(self):
        return self.salario / 2

equipa = [Funcionario("Rui", 1000), Gestor("Marta", 1500, 300), Estagiario("Tiago", 800)]
total = 0
for f in equipa:
    print(f"{f.nome} : {f.pagamento():.2f}")
    total += f.pagamento()
print(f"Total a pagar: {total:.2f} €")
```
