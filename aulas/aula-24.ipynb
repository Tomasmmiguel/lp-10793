# Exercício 1: completa o gráfico de linhas

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)

df["receita"] = df["preco"] * df["quantidade"]
receita = df.groupby("semana")["receita"].sum()
receita.plot(kind="line", marker="o")
plt.title("Receita por semana")
plt.xlabel("Semana")
plt.ylabel("Euros")
plt.show()
print(receita.round(2))
```

# Exercício 2: o gráfico de barras da Água

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)
agua = df[df["produto"] == "Água"]
vendas = agua.groupby("semana")["quantidade"].sum()
vendas.plot(kind="line", marker="o")
plt.title("Água vendida por semana")
plt.xlabel("Semana")
plt.ylabel("Quantidade")
plt.show()
print(vendas)
```

# Atividade: os gráficos da cantina

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df["receita"] = df["preco"] * df["quantidade"]
por_semana = df.groupby("semana")["receita"].sum()
por_semana.plot(kind="line", marker="o")
plt.title("Receita por semana")
plt.xlabel("Semana")
plt.ylabel("Euros")
plt.savefig("receita_semana.png")
plt.show()
por_categoria = df.groupby("categoria")["receita"].sum()
por_categoria.plot(kind="bar")
plt.title("Receita por categoria")
plt.xlabel("Categoria")
plt.ylabel("Euros")
plt.tight_layout()
plt.savefig("receita_categoria.png")
plt.show()
print(por_semana)
print(por_categoria)
```

# Desafio: os produtos mais vendidos

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)

total = df.groupby("produto")["quantidade"].sum().sort_values()
total.plot(kind="barh")
plt.title("Quantidade vendida por produto")
plt.xlabel("Quantidade")
plt.tight_layout()
plt.show()
print(total.idxmax(), total.max())
```
