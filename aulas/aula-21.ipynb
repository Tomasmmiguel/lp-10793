# Fase 1: a classe base Item

```python
class Item:
    def __init__(self, titulo, exemplares):
        self.titulo = titulo
        self._exemplares = exemplares

    @property
    def exemplares(self):
        return self._exemplares

    @exemplares.setter
    def exemplares(self, valor):
        if valor >= 0:
            self._exemplares = valor
        else:
            print("Exemplares não pode ser negativo.")
    
    @staticmethod
    def valido(titulo, exemplares):
        return len(titulo.strip()) >= 2 and exemplares >= 0
    
    @classmethod
    def de_dicionario(cls, dados):
        if cls.valido(dados["titulo"], dados["exemplares"]):
            return cls(dados["titulo"], dados["exemplares"])
        return None

    def disponivel(self):
        return self._exemplares >= 1

    def dias_emprestimo(self):
        return 14

    def multa(self, dias_atraso):
        return dias_atraso * 0.10

    def __str__(self):
        return f"{self.titulo} ({self.exemplares} ex.)"

    def __eq__(self, outro):
        return self.titulo.lower() == outro.titulo.lower()
```

# Fase 2: as subclasses Livro e Revista

```python
class Livro(Item):
    def __init__(self, titulo, exemplares, autor):
        super().__init__(titulo, exemplares)
        self.autor = autor

    def dias_emprestimo(self):
        return 21

    def __str__(self):
        return super().__str__() + " - " + self.autor

    @classmethod
    def de_dicionario(cls, dados):
        if cls.valido(dados["titulo"], dados["exemplares"]):
            return cls(dados["titulo"], dados["exemplares"], dados["autor"])
        return None


class Revista(Item):
    def __init__(self, titulo, exemplares, numero):
        super().__init__(titulo, exemplares)
        self.numero = numero

    def dias_emprestimo(self):
        return 7

    def multa(self, dias_atraso):
        return dias_atraso * 0.25

    def __str__(self):
        return super().__str__() + " - n.º " + str(self.numero)
```

# Fase 3: o decorador relatorio

```python
def relatorio(funcao):
    def nova():
        print("--- Catálogo ---")
        funcao()
        print("--- Fim do catálogo ---")
    return nova
```

# Programa de teste (não o alteres)

```python
catalogo = [Livro("Dom Casmurro", 3, "Machado de Assis"),
            Revista("Visão", 5, 12),
            Item("Dicionário", 2)]

dados = {"titulo": "Os Maias", "exemplares": 1, "autor": "Eça de Queirós"}
catalogo.append(Livro.de_dicionario(dados))

print(Livro.de_dicionario({"titulo": " ", "exemplares": 1, "autor": "X"}))


@relatorio
def mostrar_catalogo():
    for item in catalogo:
        print(item)


mostrar_catalogo()

total = 0
for item in catalogo:
    multa = item.multa(5)
    print(f"{item.titulo}: {item.dias_emprestimo()} dias, multa {multa:.2f} €")
    total += multa
print(f"Total das multas: {total:.2f} €")

catalogo[1].exemplares = -2
print(catalogo[1].exemplares)
print(Item("dom casmurro", 9) == catalogo[0])
print(Item.valido("A", 1))
```
