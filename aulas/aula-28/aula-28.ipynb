# Perguntas rápidas: o que escreve este código?

```python
import pandas as pd

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)
print(df["quantidade"].sum()) # Dá print na quantidade total de produtos vendidos
print(df[df["categoria"] == "Fruta"]["quantidade"].sum()) # Dá print na quantidade total de frutas vendidas
print(df.groupby("semana")["quantidade"].sum()) # Dá print na quantidade total de produtos vendidos por semana
```

# Exercício 1: a categoria mais vendida na semana 3 

```python
import pandas as pd

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)

sem3 = df[df["semana"] == 3]
total = sem3.groupby("categoria")["quantidade"].sum()
print(total)
```

# Exercício 2: gráfico de barras da receita

```python
import matplotlib.pyplot as plt

receita = df.groupby("categoria")["receita"].sum()
receita.plot(kind="bar", color="steelblue")
plt.title("Receita por categoria")
plt.xlabel("Categoria")
plt.ylabel("Euros")
plt.tight_layout()
plt.show()
print(receita.round(2))
```

# Exercício 3: gráfico de linhas do sumo de laranja

```python
import matplotlib.pyplot as plt

sumo = df[df["produto"] == "Sumo de laranja"]
por_semana = sumo.groupby("semana")["quantidade"].sum()
por_semana.plot(kind="line", marker="o")
plt.title("Sumo de laranja por semana")
plt.xlabel("Semana")
plt.ylabel("Unidades")
plt.show()
print(por_semana)
```
