# Exercício 1: prevê o resultado

```python
class Termometro:
    def __init__(self):
        self._graus = 20

    def ajustar(self, graus):
        if graus >= -10 and graus <= 40:
            self._graus = graus
        else:
            print("Fora do intervalo")

    def graus(self):
        return self._graus

t = Termometro()
t.ajustar(25)
t.ajustar(90)
print(t.graus())
```

# Exercício 2: completa a propriedade graus

```python
class Termometro:
    def __init__(self):
        self._graus = 20

    @property
    def graus(self):
        return self._graus

    @graus.setter
    def graus(self, valor):
        if valor >= -10 and valor <= 40:
            self._graus = valor

t = Termometro()
t.graus = 25
t.graus = 90
print(t.graus)
```

# Exercício 3: altera um exemplo que funciona

```python
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def __str__(self):
        return f"{self.titulo}, de {self.autor} ({self.ano})"

    

l = Livro("Mensagem", "Fernando Pessoa", 1934)
print(l)
```

# Atividade: a classe Produto

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    @property
    def preco(self):
        return self._preco

    @preco.setter
    def preco(self, valor):
        if valor > 0:
            self._preco = valor
        else:
            print("Preço inválido")
        
    def __str__(self):
        return f"{self.nome} - {self.preco:.2f} €"

    def __eq__(self, outro):
        return self.nome == outro.nome

p1 = Produto("Caneta", 1.5)
p2 = Produto("Caneta", 2.0)
p1.preco = -1
print(p1)
print(p1 == p2)
```
