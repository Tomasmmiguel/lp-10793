# Exercício 0: cria a tua base de dados

```python
import sqlite3
ligacao = sqlite3.connect("musica.db", timeout=5)
try:
    cursor = ligacao.cursor()
    cursor.execute("""CREATE TABLE IF NOT EXISTS musicas (
        id INTEGER PRIMARY KEY, titulo TEXT, artista TEXT)""")
    cursor.execute("DELETE FROM musicas")
    cursor.executemany("INSERT INTO musicas (id, titulo, artista) VALUES (?, ?, ?)",
    [(1, "Rosa", "Banda A"), (2, "Lume", "Banda B"), (3, "Mar", "Banda C")])
    ligacao.commit()
    cursor.execute("SELECT titulo, artista FROM musicas")
    for linha in cursor.fetchall():
        print(linha)
finally:
    ligacao.close()
```

# Atividade: qual é mais rápido?

```python
import random
import timeit

numeros = [random.randint(1, 5000) for _ in range(3000)]

def tem_duplicados_lista(lista):
    vistos = []

    for n in lista:
        if n in vistos:
            return True
        vistos.append(n)

    return False

def tem_duplicados_set(lista):
    vistos = set()

    for n in lista:
        if n in vistos:
            return True
        vistos.add(n)

    return False

t1 = timeit.timeit(lambda: tem_duplicados_lista(numeros), number=20)
t2 = timeit.timeit(lambda: tem_duplicados_set(numeros), number=20)


print("Com lista:", round(t1, 4), "s")
print("Com set:", round(t2, 4), "s")

if t1 > t2:
    print(f"A versão com set é cerca de {t1 / t2:.0f} vezes mais rápida.")
else:
    print(f"A versão com lista é cerca de {t2 / t1:.0f} vezes mais rápida.")

numeros = [random.randint(1, 5000) for _ in range(30000)]

t1 = timeit.timeit(lambda: tem_duplicados_lista(numeros), number=20)
t2 = timeit.timeit(lambda: tem_duplicados_set(numeros), number=20)

print("\nExtensão - 30000 números:")
print("Com lista:", round(t1, 4), "s")
print("Com set:", round(t2, 4), "s")

if t1 > t2:
    print(f"A versão com set é cerca de {t1 / t2:.0f} vezes mais rápida.")
else:
    print(f"A versão com lista é cerca de {t2 / t1:.0f} vezes mais rápida.")
```
