# Exercício 1: prevê o resultado

```python
def saudacao(lingua):
    def dizer(nome):
        if lingua == "pt":
            return f"Olá, {nome}"
        return f"Hello, {nome}"
    return dizer

pt = saudacao("pt")
en = saudacao("en")
print(pt("Ana"))
print(en("Rui"))
```

# Exercício 2: completa o decorador moldura

```python
def moldura(funcao):
    def nova():
        print("=" * 10)
        funcao()
        print("=" * 10)
    return nova

@moldura
def titulo():
    print("RELATÓRIO")

titulo()
```

# Exercício 3: altera um exemplo que funciona

```python
class Temperatura:
    def __init__(self, celsius):
        self.celsius = celsius
    
    @staticmethod
    def c_para_f(c):
        return c * 9 / 5 + 32

    @classmethod
    def de_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    @classmethod
    def de_kelvin(cls, k):
        return cls(k - 273.15)
    
t = Temperatura.de_kelvin(300)
print(f"{t.celsius:.2f}")
```

# Atividade: a classe Retangulo com decoradores

```python
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura

    @property
    def area(self):
        return self.largura * self.altura

    @staticmethod
    def valido(medida):
        return medida > 0

    @classmethod
    def quadrado(cls, lado):
        return cls(lado, lado)

r = Retangulo(4, 3)
q = Retangulo.quadrado(5)
print(r.area)
print(q.area)
print(Retangulo.valido(-2))
    
```

# Desafio 1: um decorador que limita as chamadas

```python
def limitar(funcao):
    chamadas = []
    def nova():
        if len(chamadas) < 3:
            chamadas.append(1)
            print(f"Chamada {len(chamadas)} de 3")
            funcao()
        else:
            print("Limite atingido")
    return nova

@limitar
def enviar():
    print("Mensagem enviada")

@limitar
def guardar():
    print("Dados guardados")
    
for i in range(4):
    enviar()
guardar()
```

# Desafio 2: produtos a partir de linhas de um CSV

```python
class Produto:
    def __init__(self, nome, preco, qtd):
        self.nome = nome
        self._preco = preco
        self._qtd = qtd

    @property
    def total(self):
        return self._preco * self._qtd

    @staticmethod
    def valido(preco, qtd):
        return preco > 0 and qtd > 0
    
    @classmethod
    def de_linha(cls, linha):
        preco = float(linha[1])
        qtd = int(linha[2])
        if not cls.valido(preco, qtd):
            return None
        return cls(linha[0], preco, qtd)

class ProdutoSaldo(Produto):
    @property
    def total(self):
        return self._preco * self._qtd * 0.8

linhas = [["Caneta", "1.50", "10", "normal"],
["Mochila", "25", "2", "saldo"],
["Régua", "-2", "3", "normal"],
["Estojo", "8", "1", "saldo"]]
soma = 0
for linha in linhas:
    if linha[3] == "saldo":
        p = ProdutoSaldo.de_linha(linha)
    else:
        p = Produto.de_linha(linha)
    if p is None:
        print("Ignorada:", linha[0])
    else:
        print(f"{p.nome}: {p.total:.2f} €")
        soma += p.total
print(f"Total: {soma:.2f} €")
```
