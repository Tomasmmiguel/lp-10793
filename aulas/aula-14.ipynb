# Exercício 1: prevê o resultado

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota >= 14")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
```

# Exercício 2: completa com WHERE e ORDER BY

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota < ? ORDER BY nota ASC", (14,))
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
```

# Exercício 3: altera um exemplo que funciona

```python
import sqlite3
ligacao = sqlite3.connect("escola.db", timeout=10)
cursor = ligacao.cursor()
cursor.executemany("INSERT INTO alunos (id, nome, nota) VALUES (?, ?, ?)", [(6, "Francisco", 13.0), (7, "Gabriela", 17.0)])
ligacao.commit()
cursor.execute("SELECT nome, nota FROM alunos")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
```

# Exercício 4: escreve de raiz

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
ligacao.row_factory = sqlite3.Row
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos WHERE nota >= 10 AND nota <= 14")
resultado = [dict(linha) for linha in cursor.fetchall()]
ligacao.close()
for aluno in resultado:
    print(aluno["nome"])
```

# Atividade: relatório da turma

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT COUNT(*) FROM alunos WHERE nota < 10")
reprovados = cursor.fetchone()[0]
cursor.execute("SELECT AVG(nota) FROM alunos")
media = cursor.fetchone()[0]
cursor.execute("SELECT nome FROM alunos ORDER BY nota DESC LIMIT 1")
melhor_aluno = cursor.fetchone()[0]
cursor.execute("SELECT COUNT(*) FROM alunos WHERE nota >= 17")
excelentes = cursor.fetchone()[0]

print(f"Média das notas: {media:.2f}")
print(f"Número de alunos com nota abaixo de 10: {reprovados}")
print(f"Melhor aluno: {melhor_aluno}")
print(f"Número de alunos com nota igual ou superior a 17: {excelentes}")

ligacao.close()
```
