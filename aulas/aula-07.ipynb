# Exercício 1: prevê e confirma

```python
alunos = ["Ana", "Bruno", "Carla", "Duarte", "Eva"]
print(alunos[2])
print(alunos[-1])
print(alunos[1:4])
print(len(alunos))
```

# Exercício 2: completa e prevê

```python
numeros = [10, 20, 30, 40, 50]
print(numeros[1:4]) # queremos [20, 30, 40]
print(numeros[:2]) # queremos os dois primeiros
print(numeros[-1:]) # queremos o último
print(numeros[::-1])
```

# Exercício 3: completa a lista de preços

```python
precos = [1.5, 2.0, 0.8]
precos.append(3.2) # acrescenta 3.2 no fim
precos.remove(2.0) # tira o 2.0
precos.sort() # ordena por ordem crescente
print(precos)
total = sum(precos)
print(f"Total: {total:.2f} €")
```

# Exercício 4: numera e soma

```python
precos = [1.5, 2.0, 0.8, 3.2]
for i, preco in enumerate(precos, 1):
    print(f"{i}: {preco:.2f} €")
total = sum(precos)
print(f"Total: {total:.2f} €")
```

# Atividade: gestor de lista de compras

```python
lista = []
for i in range(3):
    produto = input(f"Introduz o produto {i+1}: ")
    lista.append(produto)
remover = input("Qual produto quer remover? ")
if remover in lista:
    lista.remove(remover)
print("Lista de compras: ")
for i, produto in enumerate(lista, 1):
    print(f"{i}. {produto} €")
```
