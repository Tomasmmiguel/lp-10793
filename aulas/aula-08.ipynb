# Exercício 2: inventário com get() e items()

```python
stock = {"teclado": 10, "rato": 25, "monitor": 4}
print(stock.get("monitor"))
print(stock.get("webcam", 0))
for produto, qtd in stock.items():
    print(f"{produto}: {qtd} unidades")
```

# Atividade: gestor de inventário

```python
inventario = {}
for i in range(3):
    produto = input(f"Introduza o nome do produto {i+1}: ")
    quantidade = int(input(f"Introduza a quantidade de {produto}: "))
    inventario[produto] = quantidade
atualizado = input("Introduza o nome do produto a atualizar: ")
if inventario.get(atualizado) is not None:
    nova_quantidade = int(input(f"Introduza a nova quantidade de {atualizado}: "))
    inventario[atualizado] = nova_quantidade
else:
    print(f"Aviso: O produto '{atualizado}' não existe no inventário.")
print("Inventário atualizado:")
for produto, quantidade in inventario.items():
    print(f"{produto}: {quantidade} unidades")
```
