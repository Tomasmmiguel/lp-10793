# Exercício 1: prevê o resultado

```python
class Livro:
    def __init__(self, titulo, paginas):
        self.titulo = titulo
        self.paginas = paginas

l1 = Livro("Python Fácil", 300)
l2 = Livro("Contos", 100)
l2.paginas = 120
print(l1.titulo, l1.paginas)
print(l2.titulo, l2.paginas)
print(l1.paginas + l2.paginas)
```

# Exercício 2: completa a classe Retangulo

```python
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura

    def area(self):
        return self.largura * self.altura
    
r = Retangulo(4, 3)
print(r.area())
```

# Exercício 3: altera um exemplo que funciona

```python
class Produto:
    def __init__(self, nome, preco, quantidade):
        self.nome = nome
        self.preco = preco
        self.quantidade = quantidade

    def com_iva(self):
        return self.preco * self.quantidade

cesto = [Produto("Pão", 1.20, 10),
        Produto("Leite", 0.90, 5)]
for p in cesto:
    print(f"{p.nome}: {p.com_iva():.2f} €")
```

# Atividade: a classe Jogador

```python
class Jogador:
    def __init__(self, nome):
        self.nome = nome
        self.pontos = 0

    def ganhar(self, pontos):
        self.pontos += pontos

    def mostrar(self):
        print(f"{self.nome} tem {self.pontos} pontos.")

jogador1 = Jogador("Ana")
jogador2 = Jogador("Rui")
jogador1.ganhar(10)
jogador1.ganhar(5)
jogador2.ganhar(7)
jogador1.mostrar()
jogador2.mostrar()
        
```
