# Exercício 1: prevê o resultado

```python
n = 6
while n > 0:
    print(n)
    n = n - 2
print("Acabou com n =", n)
```

# Exercício 2: completa os espaços

```python
# Parte A — somar de 1 a 10
total = 0
for i in range(1, 11):
    total = total + i
print("Soma de 1 a 10:", total)
# Parte B — contagem 5 4 3 2 1
for contagem in range(5, 0, -1):
    print(contagem)
print("Lançar!")
```

# Problema A - Tabuada

```python
numero = int(input("Número: "))
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

# Problema B — Adivinha o número

```python
segredo = 7
tentativas = 0
while True:
    palpite = int(input("Palpite: "))
    tentativas = tentativas + 1
    if palpite == segredo:
        print(f"Acertou em {tentativas} tentativas!")
        break
    elif palpite < segredo:
        print("O segredo é maior.")
    else:
        print("O segredo é menor.")
```
