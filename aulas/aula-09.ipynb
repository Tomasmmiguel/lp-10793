# Exercício 1: prevê a função

```python
def area_retangulo(largura, altura):
    return largura * altura

def area_triangulo(base, altura, lado):
    return (base * altura) / 2

print(area_retangulo(4, 3))
resultado = area_retangulo(5, 2)
print(f"Área do retangulo: {resultado} m²")

print(area_triangulo(5, 6, 9))
print(f"Área do triangulo: {area_triangulo(3, 4, 5)} m²")
```

# Exercício 2: completa a função com desconto

```python
def preco_final(preco, iva=0.23):
    return preco + preco * iva

print(preco_final(100))
print(preco_final(100, 0.13))
print(preco_final(100, 0.06))
```

# Exercício 3: chama com argumentos nomeados

```python
def cabecalho(disciplina, turma="12ºPI", 
ano="2026/27"):
    return f"{disciplina} · {turma} · {ano}"

print(cabecalho("Linguagens de Programação"))
print(cabecalho("Linguagens de Programação",ano="2025/26"))
print(cabecalho(disciplina="LP", turma="12ºPI", ano="2026/27"))
print(cabecalho(turma="12ºPI", disciplina="Redes", ano="2026/27"))
```

# Exercício 4: junta as duas funções

```python
def media(notas):
    return sum(notas) / len(notas)

def classifica(media_final, minima=12):
    if media_final >= minima:
        return "Aprovado"
    return "Reprovado"

turma_a = [12, 8, 15]
turma_b = [16, 18, 14]
# mostra a média e a classificação de cada turma

media_a = media(turma_a)
print(f"Média da turma A: {media_a:.2f} - {classifica(media_a)}")
media_b = media(turma_b)
print(f"Média da turma B: {media_b:.2f} - {classifica(media_b)}")
```

# Atividade: calculadora de médias

```python
notas = []
for i in range(3):
    nota = float(input(f"Insira a nota {i+1}: "))
    notas.append(nota)
m= media(notas)
print(f"Média: {m:.2f} - {classifica(m)}")

def comentario(m):
    if m < 10:
        return ("É necessário estudar mais!")
    elif m < 14:
        return ("Está quase lá, continue a estudar!")
    elif m < 18:
        return ("Muito bem, continue a estudar!")
    else:
        return ("Excelente, continue a estudar!")

print("Comentário: " + comentario(m))
```
