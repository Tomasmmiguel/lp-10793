# Exercício 2: completa com DictReader

```python
import csv
with open("stock.csv", encoding="utf-8") as f:
    leitor = csv.reader(f)
    next(leitor)  # Ignora o cabeçalho
    mais_caro = None
    for linha in leitor:
        preco = float(linha[2])
        if mais_caro is None or preco > mais_caro:
            mais_caro = preco
print(f"Preço mais alto: {mais_caro:.2f} €")
```

# Exercício 3: altera um exemplo que funciona

```python
import csv
with open("stock.csv", encoding="utf-8") as f, \
    open("stock_novo.csv", "w", encoding="utf-8", newline="") as novo:
    leitor = csv.DictReader(f)
    escritor = csv.DictWriter(novo, fieldnames=["produto", "quantidade", "preco"])
    escritor.writeheader()
    for linha in leitor:
        escritor.writerow(linha)
    escritor.writerow({"produto": "Lapiseira", "quantidade": 25, "preco": 1.20})
```

# Atividade: relatório de stock

```python
import csv
import openpyxl

wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Relatório de Stock"

ws.append(["produto", "quantidade", "preco", "valor total"])

with open("stock.csv", encoding="utf-8", newline="") as f:
    leitor = csv.reader(f)
    next(leitor)  # Ignorar o cabeçalho
    for linha in leitor:
        produto = linha[0]
        quantidade = int(linha[1])
        preco = float(linha[2])
        valor_total = quantidade * preco
        ws.append([produto, quantidade, preco, valor_total])

wb.save("stock_relatorio.xlsx")
wb = openpyxl.load_workbook("stock_relatorio.xlsx")
ws = wb.active
for linha in ws.iter_rows(values_only=True):
    print(linha)
```
